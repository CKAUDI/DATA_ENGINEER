```python
# Silver Layer Removing NULL values

from pyspark.sql.functions import col

silver_df = spark.table("sales_catalog.pipeline.bronze_sales") \
    .filter(col("price").isNotNull()) \
    .withColumn("price", col("price").cast("double")).display()



silver_df.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable("sales_catalog.pipeline.silver_sales")
```


CREATE OR REPLACE TABLE sales_catalog.pipeline.silver_sales AS
SELECT 
    CAST(price AS DOUBLE) As price, 
    * EXCEPT(price)
FROM sales_catalog.pipeline.bronze_sales
WHERE PRICE IS NOT NULL;


DESCRIBE TABLE sales_catalog.pipeline.silver_sales;
