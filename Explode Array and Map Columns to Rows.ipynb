```python
# Create SparkSession and Prepare sample Data


from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Explode functions").getOrCreate()

arrayData = [
        ('James',['Java','Scala'],{'hair':'black','eye':'brown'}),
        ('Michael',['Spark','Java',None],{'hair':'brown','eye':None}),
        ('Robert',['CSharp',''],{'hair':'red','eye':''}),
        ('Washington',None,None),
        ('Jefferson',['1','2'],{})
]        

df = spark.createDataFrame(data = arrayData, schema = ['name','knownLanguages','properties'])
df.printSchema()
df.show()
```

```python
#explode() – PySpark explode array or map column to rows

#explode – array column example

from pyspark.sql.functions import explode

df2 =df.select(df.name, explode(df.knownLanguages))
df2.printSchema()
df2.show(truncate=False)
```

```python
df3 = df.select(df.name, explode(df.properties))
df3.printSchema()
df3.show()
```

```python
#explode_outer() returns NULL if no data – Create rows for each element in an array or map.

from pyspark.sql.functions import explode_outer
df.select(df.name, explode_outer(df.knownLanguages)).show()
df.select(df.name, explode_outer(df.properties)).show()
```

```python
#posexplode() will not return the col for empty and null values – explode array or map elements to rows

from pyspark.sql.functions import posexplode

df.select(df.name, posexplode(df.knownLanguages)).show()
df.select(df.name, posexplode(df.properties)).show()
```

```python
#posexplode_outer()  will return all empty Null values also – explode array or map columns to rows.
from pyspark.sql.functions import posexplode_outer
#Scala 
# df.select($"name",posexplode_outer($"knownLanguages")).show()

df.select(df.name, posexplode_outer(df.knownLanguages)).show()
df.select(df.name, posexplode_outer(df.properties)).show()
```
