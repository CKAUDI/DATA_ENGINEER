```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, count, desc, length, when, col, regexp_replace

```

```python
df1 = spark.read.format("csv").option("header","true").load("/Volumes/workspace/amazon_data/amazon/amazon.csv")
df1.display()
```

```python
df1.printSchema()
```

```python
top_rated_product = df1.groupBy('product_id','product_name').agg(avg('rating').alias('avg_rating')).orderBy(desc('avg_rating')).limit(10)

top_rated_product.display()
```

```python
#most reviewed 

most_reviewed = df1.groupBy('product_id','product_name').count().orderBy(desc('count')).limit(30)
most_reviewed.display()
```

```python
df1.display()
df1.printSchema()
```

```python
#discount_analysis

df1_clean = df1.withColumn(
    "discount_percentage_clean",
    regexp_replace(col("discount_percentage"), "%", "").cast("double")
)



discount = df1_clean.groupBy('category').agg(avg('discount_percentage_clean').alias('avg_discount')).display()
```

```python
user_engagement = df1.groupBy('product_id').agg(avg('rating').alias('avg_rating'), count('rating').alias('avg_count')).display()
```
