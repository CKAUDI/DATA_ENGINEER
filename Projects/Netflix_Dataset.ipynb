```python
from pyspark.sql.functions import col, when, count
from pyspark.sql.types import StringType, IntegerType, NumericType
```

```python
df = spark.read.format("csv").option("header","True").load("/Volumes/workspace/amazon_data/amazon/Netflix.csv")
df.display()
```

```python
df.printSchema()
```

```python
null_count = df.select([
    count(
        when(
            col(c).isNull() | 
            (col(c).isNaN() if isinstance(df.schema[c].dataType, NumericType) else False),
            c
        )
    ).alias(c)
    for c in df.columns
])
display(null_count)
```
