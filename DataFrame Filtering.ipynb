```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField
from pyspark.sql.types import ArrayType, IntegerType, StringType

#Create SparkSession Object
spark = SparkSession.builder.appName('Filtering Columns').getOrCreate()

#Create data
data = [
     (("James","","Smith"),["Java","Scala","C++"],"OH","M"),
    (("Anna","Rose",""),["Spark","Java","C++"],"NY","F"),
    (("Julia","","Williams"),["CSharp","VB"],"OH","F"),
    (("Maria","Anne","Jones"),["CSharp","VB"],"NY","M"),
    (("Jen","Mary","Brown"),["CSharp","VB"],"NY","M"),
    (("Mike","Mary","Williams"),["Python","VB"],"OH","M")
]

#Create Schema

schema = StructType([
    StructField('name', StructType([
        StructField('firstname', StringType(), True),
        StructField('middlename', StringType(), True),
        StructField('lastname', StringType(), True), 
    ])),
    StructField('language', ArrayType(StringType()), True),
    StructField('state', StringType(), True),
    StructField('gender', StringType(), True),

])

#create DataFrame

df = spark.createDataFrame(data = data , schema = schema)
df.printSchema()
df.show(truncate = False)


```

```python
#DataFrame filter() with Column Condition
#Using Equal condition

df.filter(df.state=='OH').show(truncate= False)
```

```python
#Using not equal condition

df.filter(df.state !='OH').show(truncate= False)
```

```python
#Using not equal condition

df.filter(~(df.state =='OH')).show(truncate= False)
```

```python
#Using Col() function

from pyspark.sql.functions import col

df.filter(col('state')=='OH').show(truncate = False)
```

```python
# Filtering with SQL Expression

df.filter("gender == 'M' ").show()

df.filter("gender != 'M' ").show()

df.filter("gender <> 'M' ").show()
```

```python
# Filter using OR operator
df.filter( (df.state  == "OH") | (df.state  == "NY") ) \
    .show(truncate=False)  

df.filter( (df.state  == "OH") | (df.gender  == "M") ) \
    .show(truncate=False)  
```

```python
#PySpark Filter with Multiple Conditions

#Using & , | operators

df.filter((df.state == "OH") & (df.gender == "M")).show(truncate = False)
df.filter((df.state=="OH")|(df.gender == "M")).show(truncate=False)
```

```python
#Filter Based on List Values

# Filter IS IN List values

li = ["OH","CA","DE"]
df.filter(df.state.isin(li)).show(truncate = False)

# Filter NOT IS IN List values

df.filter(~df.state.isin(li)).show()
df.filter(df.state.isin(li)==False).show()
```

```python
#Filter Based on Starts With, Ends With, Contains

# Using Startwith()
df.filter(df.state.startswith("N") & df.state.startswith("O")).show(truncate=False)
df.filter(df.state.startswith("N") | df.state.startswith("O")).show(truncate=False)

#using endswith

df.filter(df.state.endswith("H")).show()
#contains
df.filter(df.state.contains("H")).show()
```

```python
# Filtering with Regular Expression like(), rlike()

data2 = [(2,"Michael Rose"),(3,"Robert Williams"),
     (4,"Rames Rose"),(5,"Rames rose")
]

df2 = spark.createDataFrame(data = data2, schema = ["id", "name"])

# like - SQL LIKE pattern
df2.filter(df2.name.like("%rose%")).show()

# rlike - SQL LIKE pattern

df2.filter(df2.name.rlike("(?i)^*rose$")).show()
```

```python
# Using array_contains()

from pyspark.sql.functions import array_contains
df.filter(array_contains(df.language,"Java")).show(truncate=False)
```

```python
#Filtering on Nested Struct columns
# Struct condition

df.filter(df.name.lastname.like("%W%")).show()
```
