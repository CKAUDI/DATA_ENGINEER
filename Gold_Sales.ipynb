```python
from pyspark.sql.functions import sum

gold_df = spark.table("sales_catalog.pipeline.silver_sales") \
    .groupBy("product")  \
    .agg(sum("price").alias("total_sales"))

gold_df.display()


gold_df.write \
    .format("delta") \
        .mode("overwrite") \
            .option("mergeSchema", "true")  \
                .saveAsTable("sales_catalog.pipeline.gold_sales")


```


CREATE OR REPLACE TABLE sales_catalog.pipeline.silver_sales AS
SELECT 
  product,
  SUM(price) AS total_sales
FROM sales_catalog.pipeline.silver_sales
GROUP BY product;
